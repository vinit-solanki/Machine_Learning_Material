```python
df['Discount_value']=df['Discount'].str.extract(r'\$?(\d+\.?\d*)').astype(float)
```

```python
df.drop(columns=['Discount'],inplace=True)
df
```

```python
df.drop(columns=['Currency'],inplace=True)
df
```

```python
print(df['Rating'].str.startswith("Rated").sum())
print(df['Rating'].count())
print("Missing Reviews:", df[df["Rating"]==None])
```

```python
df.drop(columns=['Discount_value'],inplace=True)
df
```

```python
# Extract rating and total reviews using regex
df[['Rating_Value', 'Total_Reviews']] = df['Rating'].str.extract(r'Rated (\d+\.?\d*) out of 5.*?based on (\d+)')

# Convert to appropriate types
df['Rating_Value'] = df['Rating_Value'].astype(float)
df['Total_Reviews'] = df['Total_Reviews'].astype(float)
```
