```python
department_wise = df['Branch'].value_counts()
```

```python
print(department_wise)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
department_wise.plot(kind='bar')
plt.title('Number of Students by Department/Branch')
plt.xlabel('Department/Branch')
plt.ylabel('Number of Students')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show
```
